```python
# ── Gold Views: create dimension and fact views ──
# Catalog is passed as a widget parameter from the job (base_parameters).
dbutils.widgets.text("catalog", "retail_q_dab_dev", "Catalog")
catalog = dbutils.widgets.get("catalog")

# ── dim_customer ──
spark.sql(f"""
CREATE OR REPLACE VIEW {catalog}.retail_gold.dim_customer AS

SELECT
    id AS customer_id,
    customer_name,
    type AS customer_type,
    billing_city,
    billing_state,
    billing_country,
    phone,
    website,
    industry,
    annual_revenue,
    number_of_employees,
    description

FROM {catalog}.retail_silver.account

WHERE is_deleted = false and is_active=true
""")
print(f"Created view {catalog}.retail_gold.dim_customer")
```

```python
# ── dim_product ──
spark.sql(f"""
CREATE OR REPLACE VIEW {catalog}.retail_gold.dim_product AS

SELECT
    product_id,
    product_name,
    category,
    subcategory,
    brand,
    product_segment,
    unit_price,
    supplier_name,
    launch_date,
    updated_at

FROM {catalog}.retail_silver.product_catalog
WHERE is_active=true
""")
print(f"Created view {catalog}.retail_gold.dim_product")
```

```python
# ── fact_inventory ──
spark.sql(f"""
CREATE OR REPLACE VIEW {catalog}.retail_gold.fact_inventory AS

SELECT
    inventory_id,
    product_id,
    stock_quantity,
    reorder_level,
    inventory_status,
    warehouse_location,
    last_stock_update
FROM {catalog}.retail_silver.inventory
""")
print(f"Created view {catalog}.retail_gold.fact_inventory")
```
